```python
#what is regularization in machine learning

# regularization is a technique used in machine learning to prevent overfitting.
# it adds a penalty to the model's loss function to reduce complexity.
# it helps the model generalize better and perform well on unseen data.
# common types are l1 and l2 regularization.
```

```python
#diffrence between l1 and l2


# l1 regularization (lasso):
# it adds the absolute value of coefficients as a penalty.
# it can make some coefficients exactly zero, performing feature selection.

# l2 regularization (ridge):
# it adds the squared value of coefficients as a penalty.
# it reduces coefficient values but usually does not make them zero.
```

```python
#what is cross validation and why is it important

# cross-validation is a technique used to evaluate a machine learning model
# by dividing the dataset into multiple parts (folds) for training and testing.
# it helps check model performance, detect overfitting, and measure how well
# the model performs on unseen data.
```

```python
#explain k fold cross validation

# k-fold cross-validation is a technique used to evaluate a machine learning model.
# it divides the dataset into k equal parts (folds).
# the model is trained on k-1 folds and tested on the remaining fold.
# this process repeats k times, and the average score is calculated.
```

```python
#stratified k fold cross validation

# stratified k-fold cross-validation is a technique that divides the dataset
# into k folds while maintaining the same class distribution in each fold.
# it is mainly used for classification problems with imbalanced datasets.
# it ensures that every fold represents the original dataset properly.
```

```python
#loocv

# leave-one-out cross-validation (loocv) is a technique where one data point
# is used for testing and all remaining data points are used for training.
# this process repeats for every data point in the dataset.
# it provides maximum training data but is computationally expensive.
```

```python
#time series split

# time series split is a cross-validation technique used for time-dependent data.
# it divides data into training and testing sets based on time order.
# the model is trained on past data and tested on future data.
# it prevents data leakage and is commonly used for forecasting problems.
```

```python
#why are tree based models are less sensitive to feature scaling

# tree-based models are less sensitive to feature scaling because they split
# data based on feature values and thresholds rather than distances or magnitudes.
# therefore, scaling features does not significantly affect their performance.
# examples: decision tree, random forest, and xgboost.
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split , KFold, StratifiedKFold, cross_val_score, LeaveOneOut, TimeSeriesSplit , GridSearchCV , cross_validate
from sklearn.linear_model import Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import  mean_squared_error , mean_absolute_error , r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor 
from sklearn.svm import SVR
```

```python
df = pd.read_csv('Advanced_Regression_HousePrice_Dataset_3800 - Advanced_Regression_HousePrice_Dataset_3800.csv.csv')
```

```python
df
```

```python
df['sale_date'] = pd.to_datetime(df['sale_date'])
df['sale_year'] = df['sale_date'].dt.year
df['sale_month'] = df['sale_date'].dt.month
```

```python
x = df.drop(['house_price_inr', 'sale_date','property_id','near_school', 'near_metro','crime_rate_index'], axis=1)
y = df['house_price_inr']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
scaler = StandardScaler()
```

```python
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
y_train = scaler.fit_transform(y_train.values.reshape(-1, 1)).ravel()
y_test = scaler.transform(y_test.values.reshape(-1, 1)).ravel()
```

```python
params = {"alpha": [0.01, 0.1, 1, 10, 100, 1000]}
```

```python
ridge = GridSearchCV(Ridge(), params, cv=5)
ridge.fit(x_train, y_train)
```

```python
ridge_pred = ridge.predict(x_test)
```

```python
r_accu = r2_score(y_test , ridge_pred)
r_accu
```
